```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv("D:\AI Engineering\Machine-Learning\Pakistani House Price Prediction\pakistan-housing-data.csv.csv")
df
```

```python
cols = ["Beds","Baths"]
for c in cols:
    df[c] = pd.to_numeric(df[c], errors="coerce")
df[cols] = df[cols].astype("Int64")    
```

```python
median_bath = df["Baths"].median()
median_beds = df["Beds"].median()

df["Baths"].fillna(median_bath, inplace=True)
df["Beds"].fillna(median_beds, inplace=True)

# Round to nearest whole number
df["Baths"] = df["Baths"].round().astype(int)
df["Beds"] = df["Beds"].round().astype(int)
```

```python
df[["Society", "City", "Province"]] = df["Location"].str.split(",", expand=True)
df[["Society", "City", "Province"]] = df[["Society", "City", "Province"]].apply(lambda col: col.str.strip())
df.drop(columns=["Location"], inplace=True)
df.head()
```

```python
df.drop(columns=["Society"],inplace=True)
```

```python
le_city = LabelEncoder()
df["City_encoded"] = le_city.fit_transform(df["City"])
df.drop(columns=["City"], inplace=True)
df.head()
```

```python
le_province = LabelEncoder()
df["Province_encoded"] = le_province.fit_transform(df["Province"])
df.drop(columns=["Province"], inplace=True)
df.head()
```

```python
df["Area"] = df["Area"].str.strip()
df[["Area_Value", "Area_Unit"]] = df["Area"].str.split(" ", n=1, expand=True)
df.drop(columns=["Area"], inplace=True)
df.head()
```

```python
df["Area_Value"] = df["Area_Value"].str.replace(",", "")
df["Area_Value"] = pd.to_numeric(df["Area_Value"])
df["Area_Value"].dtype
```

```python
def convert_to_marla(row):
    if row["Area_Unit"] == "Kanal":
        return float(row["Area_Value"]) * 20        # 1 Kanal = 20 Marla
    elif row["Area_Unit"] == "Sq. Yd.":
        return float(row["Area_Value"]) / 25.2929   # 1 Marla = 25.2929 Sq. Yd.
    else:  # already Marla
        return float(row["Area_Value"])

df["Area_Marla"] = df.apply(convert_to_marla, axis=1)
df.drop(columns=["Area_Value", "Area_Unit"], inplace=True)
df.head()
```

```python
new_cols = ["City_encoded", "Province_encoded", "Area_Marla"]
remaining = [col for col in df.columns if col not in new_cols]
df = df[new_cols + remaining]
df.head()
```

```python
def convert_to_crore(val):
    val = val.strip().replace("PKR", "").strip()
    if "Crore" in val:
        return float(val.replace("Crore", "").strip())
    elif "Lakh" in val:
        return float(val.replace("Lakh", "").strip()) / 100  # 1 Crore = 100 Lakh
    
df["Price"] = df["Price"].apply(convert_to_crore)
df["Price"].dtype
```

```python

df["Price"].fillna(df["Price"].median(), inplace=True)
```
