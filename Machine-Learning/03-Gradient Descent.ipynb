```python
#data 
x_train = np.array([1.0,2.0])
y_train = np.array([200,300])
```

```python
def compute_cost(x,y,w,b):
    #number of examples
    m = x.shape[0]
    total_cost = 0 
    for i in range(m):
        f_wb = w*x[i] + b
        error = f_wb - y[i]
        squared_error = error**2
        total_cost = total_cost + error**2
    return total_cost / (2*m)

compute_cost(x_train,y_train,3,4)    
```

the gradient descent formula is w = w - alpha*d/dw * j(w,b) where j (w,b) = (f(w,b) - y[i] )*x[i]
same for b ==> b = b - alpha*d/db * j(w,b) where j (w,b) = (f(w,b) - y[i] )


```python
def compute_gradient(x, y, w, b):
    m = x.shape[0]
    dj_dw = 0
    dj_db = 0 
    for i in range(m):
        f_wb = w * x[i] + b 
        dj_dw_i = (f_wb - y[i]) * x[i]
        dj_db_i = f_wb - y[i] 
        dj_dw += dj_dw_i
        dj_db += dj_db_i 
    dj_dw = dj_dw / m 
    dj_db = dj_db / m
    return dj_dw, dj_db
```

```python
def gradient_descent(x, y, w_in, b_in, alpha, num_iters):
    """
    Performs gradient descent to fit w and b. Updates w and b by taking 
    num_iters gradient steps with learning rate alpha
    
    Args:
      x (ndarray (m,))  : Data, m examples 
      y (ndarray (m,))  : target values
      w_in,b_in (scalar): initial values of model parameters  
      alpha (float):     Learning rate
      num_iters (int):   number of iterations to run gradient descent
    
    Returns:
      w (scalar): Updated value of parameter after running gradient descent
      b (scalar): Updated value of parameter after running gradient descent
    """
    w = w_in
    b = b_in
    
    for i in range(num_iters):
        # Calculate the gradient and update the parameters
        dj_dw, dj_db = compute_gradient(x, y, w, b) 
        
        # Update Parameters using w, b, alpha and gradient
        w = w - alpha * dj_dw               
        b = b - alpha * dj_db               
    
    return w, b
```
