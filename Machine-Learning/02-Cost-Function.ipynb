```python
import numpy as np
import matplotlib.pyplot as plt 
```

```python
x_train = np.array([10,20,30,40])
y_train = np.array([120,230,350,410])
```

```python
"""
this function returns the cost and the prediction at the same time solely designed by me......
because i wanted to plot the graphs of the prediction as well .....
"""

def cost_and_prediction_function(x,y,w,b):
    m = x.shape[0]
    sum_cost = 0
    f_wb = []
    for i in range(m):
        f_wb_temp = w*x[i] + b 
        f_wb.append(f_wb_temp)
        error_squared = (f_wb_temp - y[i])**2
        sum_cost = sum_cost + error_squared
    total_cost = (1/(2*m)) * sum_cost
    return total_cost,f_wb
```

```python
cost,prediction = cost_and_prediction_function(x_train,y_train,10,10)
print(f"cost :{cost}")
print(f"prediction:{prediction}")
```

```python
plt.scatter(x_train,y_train,c="b",label="Actual Values")
plt.plot(x_train,prediction,c="r",label="Predicted Values")
plt.xlabel("independent variable")
plt.ylabel("dependent variable")
plt.legend()
plt.show()
```
