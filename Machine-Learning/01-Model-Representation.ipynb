```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x_train = np.array([1.0,2.0,4.0,4.5])
y_train = np.array([100,200,400,550])
m = x_train.shape[0]
m
```

```python
#linear regression = f_wb = w*x + b 
#let w = 100 , b=100
w = 80
b = 70
f_wb = []
for i in range(m):
    f_wb_temp = w*x_train[i] + b
    f_wb.append(f_wb_temp)
print(f_wb)    
```

```python
plt.scatter(x_train,y_train,c='b',label="actual values")
plt.plot(x_train,f_wb,c='r',label="predicted values")
plt.xlabel("Feature")
plt.ylabel("Target")
plt.legend()
plt.show()
```
