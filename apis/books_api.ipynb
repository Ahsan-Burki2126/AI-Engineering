```python
import requests
import pandas as pd 
import numpy as np
```

```python
response = requests.get("https://api.bigbookapi.com/search-books?query=books+about+wizards&api-key=444e1a6f644c415792d65b6eec65c0b8")
```

```python
data = response.json()
books = data["books"]
print(f"Type of the Books data is : {type(books)}")
print(f"Lenght of the books_data is : {len(books)}")
books[0]
```

```python

books_list = []
for book in books:
    books_list.append(book[0])

books_list

```

```python
print(f"type of book_list : {type(books_list)}")
print(f"type of book_list[0] : {type(books_list[0])}")
print(f"Book list keys : {books_list[0].keys()}")
```

```python
df = pd.DataFrame(books_list)
df
```

```python
df['rating_average'] = df['rating'].apply(
    lambda x: x.get('average') if isinstance(x, dict) else None
)

df.drop(columns=['rating'], inplace=True)
```

```python
df['authors'] = df['authors'].apply(
    lambda x: ', '.join(a['name'] for a in x) if isinstance(x, list) else None
)
```
